# SpaceX Falcon 9 - Data Wrangling

Explore the collected data, count launches per site and orbit, and create the landing outcome label `Class` (1 = success, 0 = failure).

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv')
df.head(10)
```

```python
df.isnull().sum() / len(df) * 100
```

```python
df.dtypes
```

## Launches per launch site

```python
df['LaunchSite'].value_counts()
```

## Launches per orbit

```python
df['Orbit'].value_counts()
```

## Mission outcome per orbit

```python
landing_outcomes = df['Outcome'].value_counts()
landing_outcomes
```

```python
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

```python
bad_outcomes = {'False ASDS', 'False Ocean', 'False RTLS', 'None ASDS', 'None None'}
bad_outcomes
```

## Create the landing outcome label

```python
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
df['Class'] = landing_class
df[['Class']].head(8)
```

```python
df.head(5)
```

```python
print('Success rate:', df['Class'].mean())
```
